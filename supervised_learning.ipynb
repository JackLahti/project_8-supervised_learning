**SETUP**

```python
import pandas as pd
from sklearn.model_selection import train_test_split
```

```python
df = pd.read_csv("Churn.csv")
```

**Analysis**

- Analyzing the dataframe checking with shape, info, and describe
- Checked to see the difference in Exited Column
- Checking for missing values in the dataframe
- Learned their is 909 missing values in the Tenure column

```python
print(df.shape)
print()
print(df.info())
print()
print(df.describe())
print()
# Check the exited column
print("\nValue counts for Exited:")
print(df['Exited'].value_counts())

print("\nMissing values per column:")
print(df.isna().sum())
print(f"\nTotal missing values: {df.isna().sum().sum()}")
print()
print(df['Tenure'].dtype)
print()

# Find the indices where Tenure is missing
missing_indices = df[df['Tenure'].isna()].index
print(f"First 10 indices with missing Tenure values: {missing_indices[:10].tolist()}")

# View the first few rows that have missing Tenure values
missing_rows = df[df['Tenure'].isna()]
print(missing_rows.head())
```

**Missing Data Column**

- Determined 909 rows is quite a lot of data
- Decided to keep the missing rows and added another column called HasFixedDeposit
- The missing values in the Tenure column will have a 0 in the HasFixDeposit

```python
# Create the HasFixedDeposit column
# .notna() returns True for non-missing values, False for missing values
# .astype(int) converts True to 1 and False to 0
df['HasFixedDeposit'] = df['Tenure'].notna().astype(int)

# Check the new column
print("Value counts for HasFixedDeposit:")
print(df['HasFixedDeposit'].value_counts())

# Verify the logic by checking a few examples
print("\nFirst few rows to verify:")
print(df[['Tenure', 'HasFixedDeposit']].head())

# Filter for rows where HasFixedDeposit is 0 and show first 10
print("\nFirst 10 customers without fixed deposits:")
print(df[df['HasFixedDeposit'] == 0][['Tenure', 'HasFixedDeposit']].head())
```

**Features & Target Setup**

- Will be only testing the data without missing values leaving the option for future testing with them
- Made the dataframe only include ints and floats so LogisticRegression would work
- Since the current project is focused on the churn rate, I will be dropping columns that are not necessary for the churn rate analysis
- Decided to drop all the columns that don't need to be tested RowNumber, CustomerId, HasFixedDeposit
- Created target and features focusing on the Exited column and then used the train_test_split()

```python
# creating a df that will test the data without the NaN
df_test = df[df['HasFixedDeposit'] == 1]

# Create a numeric version while keeping the original df_test
df_numeric = df_test.select_dtypes(include=['int64', 'float64'])

# Dropping unnecessary columns that will impact further tests
df_numeric = df_numeric.drop(['RowNumber', 'CustomerId', 'HasFixedDeposit'], axis=1)

# Check what columns we have now
print("Numeric columns:", df_numeric.columns.tolist())

target = df_numeric['Exited']
features = df_numeric.drop(['Exited'], axis=1)

features_temp, features_test, target_temp, target_test = train_test_split(
    features, target, test_size=0.2, random_state=777)

features_train, features_valid, target_train, target_valid = train_test_split(
    features_temp, target_temp, test_size=0.25, random_state=777)

# Check your class distribution
print("\nTraining set class distribution:")
print(target_train.value_counts())
print("\nValidation set class distribution:")
print(target_valid.value_counts())
```

**Test #2 Parameter Class Balance Tests**

- Created second model using LogisticRegression and using the class_weight parameter set to balanced
- With the parameter balanced classes the F1 is BELOW the 0.59 goal 

```python
model2 = LogisticRegression(random_state=777, class_weight='balanced', solver='liblinear')
model2.fit(features_train, target_train)
predicted_valid = model2.predict(features_valid)

print("Unique predictions:", np.unique(predicted_valid))
print("Prediction counts:", np.bincount(predicted_valid))

recall_balanced1 = recall_score(target_valid, predicted_valid)
precision_balanced1 = precision_score(target_valid, predicted_valid)
f1_balanced1 = f1_score(target_valid, predicted_valid)

# Your current results ARE the findings
print("\n=== Model Performance With Class Balancing ===")
print(f"F1 Score: {f1_balanced1}")
print(f"Precision: {precision_balanced1}")  
print(f"Recall: {recall_balanced1}")
print("\nFindings:")
print("- Model predicts are no longer only class 0 (no churn)")
print("- F1, Precision, and Recall are > 0.0")
print("- This shows the class_weight parameter is balancing out the classes")

# Check what probabilities the model is actually learning
predicted_proba = model2.predict_proba(features_valid)
probab_one_valid = predicted_proba[:, 1]
print(f"\nChurn probability range: {predicted_proba[:, 1].min():.3f} to {predicted_proba[:, 1].max():.3f}")
print(f"Mean churn probability: {predicted_proba[:, 1].mean():.3f}")
auc_roc_balanced1 = roc_auc_score(target_valid, probab_one_valid)
print("AUC_ROC:", auc_roc_balanced1)
```

**Test #3 Unsampled Tests**

- First created features_unsampled and target_unsampled using a repeat of 4
- Created third model using LogisticRegression and relying on my unsampled balancing to balance out classes
- EDIT: Fixed features_unsampled and target_unsampled to properly times the repeat to balance the data
- EDIT: Added an extra value_counts() to help display the change in data
- With the unsampled classes the F1 is BELOW the 0.59 goal 

```python
# Check your class distribution
print("\nTraining set class distribution:")
print(target_train.value_counts())

features_zeros = features_train[target_train == 0]
features_ones = features_train[target_train == 1]
target_zeros = target_train[target_train == 0]
target_ones = target_train[target_train == 1]

#looking at the class distribution, I will make the repeat 4
repeat = 4
features_ones_repeated = pd.concat([features_ones] * repeat, ignore_index=True)
target_ones_repeated = pd.concat([target_ones] * repeat, ignore_index=True)

features_unsampled = pd.concat([features_zeros, features_ones_repeated])
target_unsampled = pd.concat([target_zeros, target_ones_repeated])

print("\nUnsampled class distribution:")
print(target_unsampled.value_counts())

model3 = LogisticRegression(random_state=777, solver='liblinear')
model3.fit(features_unsampled, target_unsampled)
predicted_valid = model3.predict(features_valid)

print("\nUnique predictions:", np.unique(predicted_valid))
print("Prediction counts:", np.bincount(predicted_valid))

recall_balanced2 = recall_score(target_valid, predicted_valid)
precision_balanced2 = precision_score(target_valid, predicted_valid)
f1_balanced2 = f1_score(target_valid, predicted_valid)

# Your current results ARE the findings
print("\n=== Model Performance With Unsampled Class Balancing ===")
print(f"F1 Score: {f1_balanced2}")
print(f"Precision: {precision_balanced2}")  
print(f"Recall: {recall_balanced2}")

# Check what probabilities the model is actually learning
predicted_proba = model3.predict_proba(features_valid)
probab_one_valid = predicted_proba[:, 1]
print(f"\nChurn probability range: {predicted_proba[:, 1].min():.3f} to {predicted_proba[:, 1].max():.3f}")
print(f"Mean churn probability: {predicted_proba[:, 1].mean():.3f}")
auc_roc_balanced2 = roc_auc_score(target_valid, probab_one_valid)
print("AUC_ROC:", auc_roc_balanced2)
```

**Test #4 DecisionTreeClassifier Tests**

- Created fourth model using DecisionTreeClassifier and using class_weight='balanced'
- Testing depth till I found depth=5 to be a good one
- Tested different thresholds to get the closest f1 score to the goal of 0.59

```python
model4 = DecisionTreeClassifier(
    random_state=777, 
    class_weight='balanced', 
    max_depth=5
)
model4.fit(features_train, target_train)
predicted_valid = model4.predict(features_valid)

# Get the predicted probabilities from your model4
predicted_proba = model4.predict_proba(features_valid)
probab_one_valid = predicted_proba[:, 1]  # Probabilities for class 1 (churn)
auc_roc_tree = roc_auc_score(target_valid, probab_one_valid)

# Test different thresholds
thresholds_to_test = [0.35, 0.40, 0.45, 0.48, 0.50, 0.52, 0.55, 0.60]

best_f1 = 0
best_threshold = 0.5


print("Threshold Tuning Results:")
print("=" * 50)

best_precision = 0
best_recall = 0
best_precision_threshold = 0.5
best_recall_threshold = 0.5

for threshold in thresholds_to_test:
    predicted_threshold = (probab_one_valid >= threshold).astype(int)
    
    # Calculate all three metrics
    f1_threshold = f1_score(target_valid, predicted_threshold)
    precision_threshold = precision_score(target_valid, predicted_threshold)
    recall_threshold = recall_score(target_valid, predicted_threshold)
    
    print(f"Threshold: {threshold:.2f} | F1: {f1_threshold:.4f} | Precision: {precision_threshold:.4f} | Recall: {recall_threshold:.4f}")
    
    # Track best F1 (you already have this)
    if f1_threshold > best_f1:
        best_f1 = f1_threshold
        best_threshold = threshold
    
    # Track best precision
    if precision_threshold > best_precision:
        best_precision = precision_threshold
        best_precision_threshold = threshold
        
    # Track best recall
    if recall_threshold > best_recall:
        best_recall = recall_threshold
        best_recall_threshold = threshold

print("=" * 50)
print(f"Best Threshold: {best_threshold}")
print(f"Best F1 Score: {best_f1:.4f}")

print("\n=== #4 Model Performance With Unsampled Class Balancing ===")
print(f"F1 Score: {best_f1}")
print("AUC_ROC:", auc_roc_tree)
print(f"Precision: {best_precision}")  
print(f"Recall: {best_recall}")
```

**Final Test**

- Taking all the best information I gathered from the DecisionTreeClassifier I tested it using the final set
- keeping the same depth=5, the same threshold=0.6, and making sure the class_weight is balanced
- I got a final f1 score of 0.59

```python
final_model = DecisionTreeClassifier(max_depth=5, class_weight='balanced', random_state=777)
final_model.fit(features_temp, target_temp)

# Get predictions on the test set using your final_model
final_predictions_proba = final_model.predict_proba(features_test)
final_probab_one = final_predictions_proba[:, 1]

# Apply your optimal threshold (0.6) that you found during validation
optimal_threshold = 0.6
final_predictions = (final_probab_one >= optimal_threshold).astype(int)

# Calculate final metrics
final_f1 = f1_score(target_test, final_predictions)
final_auc_roc = roc_auc_score(target_test, final_probab_one)
final_precision = precision_score(target_test, final_predictions)
final_recall = recall_score(target_test, final_predictions)

print("=== FINAL TEST RESULTS ===")
print(f"F1 Score: {final_f1:.4f}")
print(f"AUC-ROC: {final_auc_roc:.4f}")
print(f"Precision: {final_precision:.4f}")
print(f"Recall: {final_recall:.4f}")
print(f"Threshold used: {optimal_threshold}")

# Check if you met the requirement
if final_f1 >= 0.59:
    print("\n🎉 SUCCESS! F1 score meets the requirement of ≥ 0.59")
else:
    print(f"\n❌ F1 score ({final_f1:.4f}) is below the requirement of 0.59")
```
